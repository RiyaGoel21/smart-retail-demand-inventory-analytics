```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("outputs/cleaned_data.csv")

inventory_df = pd.read_csv("outputs/inventory_analysis.csv")

df.head()
```

```python
df["revenue"] = (
    df["units_sold"]
    *
    df["price"]
)
```

```python
df["forecast_error"] = (
    df["units_sold"]
    -
    df["demand_forecast"]
)
```

```python
df["forecast_accuracy"] = (
    1 -
    (
        abs(df["forecast_error"])
        /
        (df["units_sold"] + 1)
    )
) * 100
```

```python
dashboard_df = pd.merge(
    df,
    inventory_df[
        [
            "product_id",
            "abc_class",
            "risk_status"
        ]
    ],
    on="product_id",
    how="left"
)
```
